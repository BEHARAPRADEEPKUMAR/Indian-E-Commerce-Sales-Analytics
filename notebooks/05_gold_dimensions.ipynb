```python
# Purpose:
# This notebook creates the Dimension Tables for the Gold layer.
#
# Design Approach:
# We are following a STAR SCHEMA design.

# The central table will be fact_sales

# The supporting Dimension Tables will be:
#     dim_customer -> Customer information (WHO)
#     dim_product  -> Product information (WHAT)
#     dim_date     -> Date/Time information (WHEN)
#
# Star Schema Design:
#
#                       dim_customer
#                            |
#                            |
#                            ↓
# dim_date  ------------ fact_sales ------------ dim_product

# Why are we creating Dimension Tables?
# 1. To separate descriptive information from transaction data.
# 2. To make the Gold layer easier to query and analyze.
# 3. To avoid repeatedly deriving customer, product and date
#    attributes in every analytics query.
# 4. To create a reusable dimensional model for reporting.

# Dimension Responsibilities:

# dim_customer:
# Stores customer-related attributes such as name, gender,
# age, city, state and customer tier.

# dim_product:
# Stores product-related attributes such as product name,
# category, brand, price and rating.

# dim_date:
# Stores calendar attributes such as date, year, month,
# month name and quarter for time-based analysis.

# These dimensions will later be connected to fact_sales
# using their respective keys.

# Data Flow:

# Silver Tables
#      ↓
# Create Gold Dimensions
#      ↓
# dim_customer
# dim_product
# dim_date
#      ↓
# Used by fact_sales and Gold Analytics
```

```python
#Reading from silver tables
from pyspark.sql import functions as F
customers = spark.table(
    "workspace.indian_ecommerce_sales_analytics.silver_customers"
)

products = spark.table(
    "workspace.indian_ecommerce_sales_analytics.silver_products"
)

sales = spark.table(
    "workspace.indian_ecommerce_sales_analytics.silver_sales"
)

print("Customers:", customers.count())
print("Products :", products.count())
print("Sales    :", sales.count())
```

```python
dim_customer = customers.select(
    "Customer_ID",
    "Customer_Name",
    "Gender",
    "Age",
    "Age_Group",
    "Date_of_Birth",
    "Email",
    "Phone",
    "City",
    "State",
    "Pincode",
    "Registration_Date",
    "Customer_Tier"
)
```

```python
print("dim_customer rows:", dim_customer.count())

print(
    "Distinct Customer IDs:",
    dim_customer.select("Customer_ID").distinct().count()
)
display(dim_customer.limit(10))
```

```python
dim_product = products.select(
    "Product_ID",
    "Product_Name",
    "Category",
    "Brand",
    "Original_Price",
    "Discount_Percent",
    "Discount_Amount",
    "Selling_Price",
    "Weight_kg",
    "Avg_Rating",
    "Total_Reviews"
)

print("dim_product rows:", dim_product.count())

print(
    "Distinct Product IDs:",
    dim_product.select("Product_ID").distinct().count()
)
```

```python
display(dim_product.limit(10))
```

```python
#Findouting sales max ,min dates
date_range = sales.select(
    F.min("Order_Date").alias("min_date"),
    F.max("Order_Date").alias("max_date")
)

display(date_range)
```

```python
date_range_values = date_range.collect()[0]

min_date = date_range_values["min_date"]
max_date = date_range_values["max_date"]

print("Minimum Order Date:", min_date)
print("Maximum Order Date:", max_date)
```

```python
dim_date = (
    spark.range(
        0,
        (max_date - min_date).days + 1
    )
    .withColumn(
        "Date",
        F.date_add(
            F.lit(min_date),
            F.col("id").cast("int")
        )
    )
    .drop("id")
)
```

```python
dim_date = (
    dim_date

    .withColumn(
        "Date_Key",
        F.date_format("Date", "yyyyMMdd").cast("int")
    )

    .withColumn(
        "Year",
        F.year("Date")
    )

    .withColumn(
        "Month",
        F.month("Date")
    )

    .withColumn(
        "Month_Name",
        F.date_format("Date", "MMMM")
    )

    .withColumn(
        "Quarter",
        F.quarter("Date")
    )

    .withColumn(
        "Quarter_Name",
        F.concat(
            F.lit("Q"),
            F.quarter("Date")
        )
    )

    .withColumn(
        "Year_Month",
        F.date_format("Date", "yyyy-MM")
    )

    .withColumn(
        "Day",
        F.dayofmonth("Date")
    )

    .withColumn(
        "Day_of_Week",
        F.date_format("Date", "EEEE")
    )

    .withColumn(
        "Week_of_Year",
        F.weekofyear("Date")
    )

    .withColumn(
        "Is_Weekend",
        F.dayofweek("Date").isin([1, 7])
    )
)
```

```python
display(
    dim_date.orderBy("Date").limit(20)
)
```

```python
#Validating date dimentions
print("Date Dimension Rows:", dim_date.count())

print(
    "Distinct Dates:",
    dim_date.select("Date").distinct().count()
)
```
