```python
silver_customers = spark.table(
    "workspace.indian_ecommerce_sales_analytics.bronze_customers"
)

display(silver_customers)
```

```python
silver_customers.printSchema()
```

```python
print("Customer Row Count:", silver_customers.count())

display(
    silver_customers.describe()
)
```

Clean String Columns

```python
from pyspark.sql import functions as F
from pyspark.sql.window import Window
string_columns = [
    "Customer_ID",
    "Customer_Name",
    "Gender",
    "Age_Group",
    "Email",
    "Phone",
    "City",
    "State",
    "Customer_Tier"
]
#remove unnecessary leading/trailing spaces.
for col_name in string_columns:
    silver_customers = silver_customers.withColumn(
        col_name,
        F.trim(F.col(col_name))
    )
```

```python
silver_customers = silver_customers.withColumn(
    "Gender",
    F.initcap(F.lower(F.col("Gender")))
)
display(
    silver_customers
    .groupBy("Gender")
    .count()
    .orderBy(F.desc("count"))
)
```

```python
silver_customers = silver_customers.withColumn(
    "Customer_Tier",
    F.upper(F.trim(F.col("Customer_Tier")))
)
display(
    silver_customers
    .groupBy("Customer_Tier")
    .count()
    .orderBy(F.desc("count"))
)
```

```python
#Applying proper data types
silver_customers = (
    silver_customers
    .withColumn("Age", F.col("Age").cast("int"))
    .withColumn("Pincode", F.col("Pincode").cast("int"))
    .withColumn("Total_Orders", F.col("Total_Orders").cast("int"))
    .withColumn("Total_Spent", F.col("Total_Spent").cast("double"))
)
```

```python
#Converting Registration Date
silver_customers = silver_customers.withColumn(
    "Registration_Date",
    F.to_date(F.col("Registration_Date"))
)
silver_customers = silver_customers.withColumn(
    "Date_of_Birth",
    F.to_date(F.col("Date_of_Birth"))
)
silver_customers.select(
    "Customer_ID",
    "Date_of_Birth",
    "Registration_Date"
    
).show(10, False)
```

```python
null_report = silver_customers.select([
    F.sum(
        F.when(F.col(c).isNull(), 1).otherwise(0)
    ).alias(c)
    for c in silver_customers.columns
])

display(null_report)
```

```python
#Checking duplicate customer IDs
duplicate_customers = (
    silver_customers
    .groupBy("Customer_ID")
    .count()
    .filter(F.col("count") > 1)
)

display(duplicate_customers)
print(
    "Duplicate Customer IDs:",
    duplicate_customers.count()
)
```

```python
#Checking Invalid age
invalid_age = silver_customers.filter(
    (F.col("Age") < 0) |
    (F.col("Age") > 120)
)

print("Invalid Age Records:", invalid_age.count())

display(invalid_age)
```

```python
#Checking negitive customer metrics
invalid_metrics = silver_customers.filter(
    (F.col("Total_Orders") < 0) |
    (F.col("Total_Spent") < 0)
)

print(
    "Invalid Customer Metric Records:",
    invalid_metrics.count()
)

display(invalid_metrics)
```

```python
#Adding Data Quality Flags
silver_customers = (
    silver_customers

    .withColumn(
        "dq_invalid_age",
        F.when(
            (F.col("Age") < 0) |
            (F.col("Age") > 120),
            True
        ).otherwise(False)
    )

    .withColumn(
        "dq_negative_orders",
        F.when(
            F.col("Total_Orders") < 0,
            True
        ).otherwise(False)
    )

    .withColumn(
        "dq_negative_spent",
        F.when(
            F.col("Total_Spent") < 0,
            True
        ).otherwise(False)
    )
)
```

```python
silver_customers = silver_customers.select(
    "Customer_ID",
    "Customer_Name",
    "Gender",
    "Age",
    "Age_Group",
    "Date_of_Birth",
    "Email",
    "Phone",
    "City",
    "State",
    "Pincode",
    "Registration_Date",
    "Customer_Tier",
    "Total_Orders",
    "Total_Spent",
    "dq_invalid_age",
    "dq_negative_orders",
    "dq_negative_spent"
)
```

```python
silver_customers.printSchema()
```

```python
print("Silver Customer Records:", silver_customers.count())

print(
    "Distinct Customer IDs:",
    silver_customers.select("Customer_ID").distinct().count()
)
```
